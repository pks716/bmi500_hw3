'''Problem 1. Write a function in Python 3.x that, given an integer n, returns the sum of
the first n terms of the series in the Leibniz formula.'''

```python

def leibniz_sum(n):
    total = 0
    for k in range(n):
        total += (-1)**k / (2*k + 1)
    return total

start_time1 = time.time()
result1 = leibniz_sum(n)
end_time1 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result1} in time {end_time1-start_time1}.")
```

'''Problem 2. For each of the following items, write a function in Python 3.x that, given an
integer n, returns the sum of the first n terms of the series in the Leibniz formula.'''

```python
'''a. Use a for-loop and an if-statement with the modulo operator % to determine whether
to add or subtract each term.'''

def leibniz_sum(n):
    total = 0
    for k in range(n):
        if k % 2 == 0:
            total += 1 / (2*k + 1)
        else:
            total -= 1 / (2*k + 1)
    return total

start_time2 = time.time()
result2 = leibniz_sum(n)
end_time2 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result2} in time {end_time2-start_time2}.")
```

```python
'''b. Use a for-loop with the quantity (-1)**n to determine whether to add or subtract
each term.'''

def leibniz_sum(n):
    total = 0
    for k in range(n):
        total += (-1)**k / (2*k + 1)
    return total

start_time3 = time.time()
result3 = leibniz_sum(n)
end_time3 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result3} in time {end_time3-start_time3}.")
```

```python
'''c. Construct a Python list and compute the sum of the terms in the list.'''

def leibniz_sum(n):
    terms = []
    for k in range(n):
        terms.append((-1)**k / (2*k + 1))
    return sum(terms)

start_time4 = time.time()
result4 = leibniz_sum(n)
end_time4 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result4} in time {end_time4-start_time4}.")
```

```python
'''d. Construct a Python set and compute the sum of the terms in the set.'''

def leibniz_sum(n):
    terms = set()
    for k in range(n):
        terms.add((-1)**k / (2*k + 1))
    return sum(terms)

start_time5 = time.time()
result5 = leibniz_sum(n)
end_time5 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result5} in time {end_time5-start_time5}.")
```

```python
'''e. Construct a Python dictionary and compute the sum of the terms in the dictionary.'''

def leibniz_sum(n):
    terms = {}
    for k in range(n):
        terms[k] = (-1)**k / (2*k + 1)
    return sum(terms.values())

start_time6 = time.time()
result6 = leibniz_sum(n)
end_time6 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result6} in time {end_time6-start_time6}.")
```

```python
'''f. Construct a NumPy array and compute the sum of the terms in the array.'''

import numpy as np

def leibniz_sum(n):
    k = np.arange(n)
    terms = (-1)**k / (2*k + 1)
    return np.sum(terms)

start_time7 = time.time()
result7 = leibniz_sum(n)
end_time7 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result7} in time {end_time7-start_time7}.")
```

```python
'''g. Construct a NumPy array, use array indexing to compute the sum of the positive terms
in the array, use array indexing to compute the sum of the negative terms in the array,
and add the two sums together. You can write x[::2] to access the first, third, etc.
terms and x[1::2] to access the second, fourth, etc. terms.'''

import numpy as np

def leibniz_sum(n):
    k = np.arange(n)
    terms = (-1)**k / (2*k + 1)
    positive_sum = np.sum(terms[::2])
    negative_sum = np.sum(terms[1::2])
    return positive_sum + negative_sum

start_time8 = time.time()
result8 = leibniz_sum(n)
end_time8 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result8} in time {end_time8-start_time8}.")
```

```python
'''j. Combine the first and second terms, the third and fourth terms, etc. to change this
series from an alternating to a non-alternating series and compute the sum of the
combined terms.'''

def leibniz_sum(n):
    total = 0
    num_pairs = n // 2
    for k in range(num_pairs):
        total += 1/(4*k + 1) - 1/(4*k + 3)
    if n % 2 == 1:
        total += (-1)**(n-1) / (2*(n-1) + 1)
    return total

start_time9 = time.time()
result9 = leibniz_sum(n)
end_time9 = time.time()
print(f"The sum of the first {n} terms of the Leibniz formula is {result9} in time {end_time9-start_time9}.")
```

'''Problem 4. Choose your favorite implementation of this formula, and plot the absolute
error in the sum as a function of the number of terms in the sum.
You should have accurate, informative, and legible labels for both the x-axis and y-axis, and
you should use logarithmic axes or take the logarithm of the errors and the numbers of terms
so that you can see the sums converge over a wide range of values.
By default, most plotting packages generate large plots that are difficult to read in papers
and presentations. Use a command like plt.figure(figsize=(2, 2)) to make your plot
more legible.'''

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def leibniz_sum(n):
    k = np.arange(n)
    terms = (-1)**k / (2*k + 1)
    return np.sum(terms)

n_values = np.logspace(0, 6, 50).astype(int)
errors = [abs(4 * leibniz_sum(n) - math.pi) for n in n_values]

plt.figure(figsize=(4, 4))
plt.loglog(n_values, errors)
plt.xlabel("Number of terms (n)")
plt.ylabel("Absolute error")
plt.title("Leibniz series convergence")
plt.grid(True, which="major", alpha=0.3)
plt.show()
```
